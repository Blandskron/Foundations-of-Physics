## 1. Notación Matemática Esencial y Análisis Combinatorio

Para abordar el modelado de datos y el cálculo de probabilidades, es imprescindible dominar dos herramientas matemáticas transversales: la **notación de sumatoria/productoria** y los **métodos de conteo combinatorio**.

### 1.1 Notación de Sumatoria ($\sum$) y Productoria ($\prod$)

Sea $x_1, x_2, \dots, x_n$ una secuencia de $n$ observaciones numéricas. La suma de todas las observaciones se denota mediante el operador sigma ($\sum$):
$$\sum_{i=1}^{n} x_i = x_1 + x_2 + \dots + x_n$$

#### Propiedades Algebraicas de la Sumatoria:
- **Linealidad con constante:** \(\sum_{i=1}^{n} (c \cdot x_i) = c \sum_{i=1}^{n} x_i\), para cualquier constante $c \in \mathbb{R}$.
- **Suma de constantes:** \(\sum_{i=1}^{n} c = n \cdot c\).
- **Distributividad respecto a la adición:** \(\sum_{i=1}^{n} (x_i + y_i) = \sum_{i=1}^{n} x_i + \sum_{i=1}^{n} y_i\).
- **Separación de desviaciones respecto al centro:** \(\sum_{i=1}^{n} (x_i - \bar{x}) = 0\), donde \(\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i\).

De manera análoga, el producto de $n$ términos se expresa con el operador pi ($\prod$):
$$\prod_{i=1}^{n} x_i = x_1 \cdot x_2 \cdots x_n$$

### 1.2 Análisis Combinatorio y Conteo Discreto

El conteo de configuraciones posibles en experimentos aleatorios discretos requiere de tres construcciones fundamentales:

1. **Factorial ($n!$):** Representa el número de formas de ordenar $n$ objetos distintos en una fila:
   $$n! = \prod_{i=1}^{n} i = n \cdot (n-1) \cdot (n-2) \cdots 1, \quad \text{con } 0! = 1$$

2. **Permutaciones ($P(n, k)$):** Número de formas de seleccionar y ordenar $k$ elementos a partir de un conjunto de $n$ elementos distintos ($k \le n$):
   $$P(n, k) = \frac{n!}{(n-k)!}$$

3. **Combinaciones / Coeficiente Binomial ($\binom{n}{k}$):** Número de formas de seleccionar $k$ elementos de un conjunto de $n$ elementos sin importar el orden de selección:
   $$\binom{n}{k} = C(n, k) = \frac{n!}{k!(n-k)!}$$


```python
import math
import numpy as np
import scipy.special as sp

# Implementación artesanal de operadores combinatorios
def factorial_artesanal(n: int) -> int:
    """Calcula el factorial de n iterativamente."""
    if n < 0:
        raise ValueError("El factorial no está definido para números negativos.")
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

def permutaciones_artesanal(n: int, k: int) -> int:
    """Calcula P(n, k) = n! / (n - k)!"""
    if k < 0 or k > n:
        return 0
    return factorial_artesanal(n) // factorial_artesanal(n - k)

def combinaciones_artesanal(n: int, k: int) -> int:
    """Calcula C(n, k) = n! / (k! * (n - k)!)"""
    if k < 0 or k > n:
        return 0
    return factorial_artesanal(n) // (factorial_artesanal(k) * factorial_artesanal(n - k))

# Caso de prueba numérico: Selección de k=3 partículas de un sistema de n=10 partículas
n_test, k_test = 10, 3

p_art = permutaciones_artesanal(n_test, k_test)
c_art = combinaciones_artesanal(n_test, k_test)

# Validación contra bibliotecas estándar (math y scipy)
p_ref = math.perm(n_test, k_test)
c_ref = math.comb(n_test, k_test)
c_scipy = int(sp.comb(n_test, k_test))

print(f"=== Validación del Análisis Combinatorio (n={n_test}, k={k_test}) ===")
print(f"Permutaciones  - Artesanal: {p_art} | Referencia (math.perm): {p_ref}")
print(f"Combinaciones  - Artesanal: {c_art} | Referencia (math.comb): {c_ref} | SciPy: {c_scipy}")

# Verificación analítica mediante assertions
assert p_art == p_ref, "Error en el cálculo de permutaciones artesanal"
assert c_art == c_ref == c_scipy, "Error en el cálculo de combinaciones artesanal"
print("[OK] Validación exitosa: Los algoritmos artesanales coinciden exactamente con la biblioteca estándar.")
```

## 2. Teoría de Conjuntos y Álgebra de Eventos para Probabilidad

La teoría de probabilidad moderna (axiomatizada por Andrey Kolmogorov en 1933) se construye sobre el marco formal de la **teoría de conjuntos**.

### 2.1 Conceptos Fundamentales

- **Espacio Muestral ($\Omega$):** El conjunto de todos los resultados posibles de un experimento aleatorio.
- **Evento o Suceso ($A \subseteq \Omega$):** Cualquier subconjunto del espacio muestral.
- **Evento Elemental ($\omega \in \Omega$):** Un único resultado indivisible del experimento.
- **Evento Imposible ($\emptyset$):** El conjunto vacío, un evento que nunca puede ocurrir.

### 2.2 Operaciones con Eventos y sus Propiedades

Dado dos eventos $A, B \subseteq \Omega$:

1. **Unión ($A \cup B$):** Evento que ocurre si ocurre $A$, ocurre $B$, o ambos.
2. **Intersección ($A \cap B$):** Evento que ocurre únicamente si $A$ y $B$ ocurren simultáneamente.
3. **Complemento ($A^c$ o $\bar{A}$):** Evento que ocurre si $A$ **no** ocurre: $A^c = \{ \omega \in \Omega \mid \omega \notin A \}$.
4. **Eventos Mutuamente Excluyentes (Disjuntos):** Dos eventos son disjuntos si no pueden ocurrir simultáneamente:
   $$A \cap B = \emptyset$$

#### Leyes de De Morgan:
Establecen la relación dual entre unión, intersección y complementación:
$$(A \cup B)^c = A^c \cap B^c$$
$$(A \cap B)^c = A^c \cup B^c$$


```python
# Representación de un espacio muestral discreto en Python mediante la estructura de datos 'set'
# Experimento: Lanzamiento de un dado de 12 caras (dado dodecaédrico)
omega = set(range(1, 13))

# Definición de eventos
# Evento A: Obtener un número par
evento_A = {x for x in omega if x % 2 == 0}
# Evento B: Obtener un número mayor o igual a 7
evento_B = {x for x in omega if x >= 7}

# Operaciones algebraicas de conjuntos
union_AB = evento_A.union(evento_B)
interseccion_AB = evento_A.intersection(evento_B)
complemento_A = omega.difference(evento_A)
complemento_B = omega.difference(evento_B)

# Demostración numérica de las Leyes de De Morgan
# Ley 1: (A U B)^c == A^c Intersección B^c
izq_ley1 = omega.difference(union_AB)
der_ley1 = complemento_A.intersection(complemento_B)

# Ley 2: (A Intersección B)^c == A^c U B^c
izq_ley2 = omega.difference(interseccion_AB)
der_ley2 = complemento_A.union(complemento_B)

print("=== Análisis de Eventos y Leyes de De Morgan ===")
print(f"Espacio Muestral Omega = {omega}")
print(f"Evento A (Pares): {evento_A}")
print(f"Evento B (>=7): {evento_B}")
print(f"Union (A U B): {union_AB}")
print(f"Intersección (A n B): {interseccion_AB}")
print(f"Leyes de De Morgan 1: (A U B)^c = {izq_ley1} | A^c n B^c = {der_ley1}")
print(f"Leyes de De Morgan 2: (A n B)^c = {izq_ley2} | A^c U B^c = {der_ley2}")

assert izq_ley1 == der_ley1, "Fallo en la verificación de la 1ra Ley de De Morgan"
assert izq_ley2 == der_ley2, "Fallo en la verificación de la 2da Ley de De Morgan"
print("[OK] Verificación formal: Las Leyes de De Morgan se cumplen exactamente en la simulación.")
```

## 3. Entorno Computacional y Manejo de Datos en Python

En la física contemporánea y el análisis de datos masivos (Big Data), el cómputo estadístico se apoya en tres librerías estándar del ecosistema Python:
- **`NumPy`**: Operaciones matriciales, vectorización de algoritmos y generación de números pseudoaleatorios.
- **`SciPy`**: Cálculo de funciones de densidad, distribuciones probabilísticas e inferencia.
- **`Matplotlib`**: Graficación científica de alta calidad y visualización de datos.

A continuación, implementaremos una simulación de **muestreo estocástico** para comparar una población teórica con datos empíricos observados.


```python
import numpy as np
import matplotlib.pyplot as plt

# Aplicar estilo científico institucional
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# Configuración de simulación estocástica reproducible
rng = np.random.default_rng(seed=42)

# Generar una población sintética de mediciones físicas (ej. velocidades moleculares en m/s)
mu_teorica = 300.0  # Velocidad media
sigma_teorica = 25.0  # Desviación típica
N_muestras = 5000

# Muestra simulada a partir de una distribución Gaussian / Normal
datos_medidos = rng.normal(loc=mu_teorica, scale=sigma_teorica, size=N_muestras)

# Cálculo de estadísticas descriptivas con NumPy
media_empirica = np.mean(datos_medidos)
desv_empirica = np.std(datos_medidos, ddof=1)  # Corrección insesgada de Bessel (ddof=1)
mediana_empirica = np.median(datos_medidos)

# Creación del gráfico de visualización estadística
fig, ax = plt.subplots(figsize=(10, 6))

# Histogramas de frecuencia relativa (density=True)
n_bins, bins, patches = ax.hist(datos_medidos, bins=35, density=True, alpha=0.6,
                               color='skyblue', edgecolor='navy', label='Frecuencia Empírica (Muestra)')

# Superposición de la curva teórica Gaussiana
x_teorico = np.linspace(mu_teorica - 4*sigma_teorica, mu_teorica + 4*sigma_teorica, 500)
pdf_teorica = (1.0 / (sigma_teorica * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_teorico - mu_teorica) / sigma_teorica)**2)
ax.plot(x_teorico, pdf_teorica, 'r-', linewidth=2.5, label=r'Densidad de Probabilidad Teórica $N(\mu, \sigma)$')

# Líneas de referencia para la media y mediana
ax.axvline(media_empirica, color='darkgreen', linestyle='--', linewidth=2, label=f'Media Muestral: {media_empirica:.2f} m/s')
ax.axvline(mediana_empirica, color='purple', linestyle=':', linewidth=2, label=f'Mediana Muestral: {mediana_empirica:.2f} m/s')

# Formato y rotulación formal
ax.set_title('Simulación de Mediciones Experimentales: Muestreo Estocástico Gaussiano', fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Velocidad Molecular (m/s)', fontsize=11)
ax.set_ylabel(r'Densidad de Frecuencia $f(x)$', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.7)
ax.legend(frameon=True, facecolor='white', framealpha=0.9, fontsize=10)

plt.tight_layout()
plt.show()

print(f"=== Parámetros Teóricos vs Estadísticos Estimados ===")
print(f"Media Teórica:   {mu_teorica:.2f} m/s  | Media Empírica:   {media_empirica:.2f} m/s (Error: {abs(media_empirica-mu_teorica):.3f})")
print(f"Desv. Teórica:    {sigma_teorica:.2f} m/s  | Desv. Empírica:    {desv_empirica:.2f} m/s (Error: {abs(desv_empirica-sigma_teorica):.3f})")
```

## 5. Reto Diagnóstico y Autoevaluación Práctica

Para evaluar tu preparación previa antes de avanzar a la **Lección 1**, completa el siguiente desafío práctico integrador en Python.

### Consigna del Desafío:
Supongamos un detector de partículas experimental con $n = 8$ sensores independientes. Queremos evaluar el número de formas de seleccionar $k = 4$ sensores para una prueba de calibración, y simular $N = 10\,000$ eventos aleatorios de detección en los que cada sensor tiene una probabilidad $p = 0.5$ de activarse.

Debes programar:
1. El cálculo exacto de las combinaciones $\binom{8}{4}$.
2. La simulación de los $N = 10\,000$ eventos con `NumPy` contando el número de sensores activos por evento.
3. La verificación de que la frecuencia empírica con que se activan exactamente 4 sensores se aproxima al valor teórico $\binom{8}{4} / 2^8$.


```python
# === SOLUCIÓN AL RETO DIAGNÓSTICO ===
import math
import numpy as np

n_sensores = 8
k_objetivo = 4
prob_sensor = 0.5
N_simulaciones = 10000

# 1. Cálculo del número de formas de seleccionar k sensores (Combinaciones)
formas_seleccion = math.comb(n_sensores, k_objetivo)

# 2. Probabilidad teórica exacta de que se activen exactamente k=4 sensores de n=8 (Modelo Binomial básico)
prob_teorica_k4 = formas_seleccion * (prob_sensor ** n_sensores)

# 3. Simulación de Monte Carlo con NumPy
rng = np.random.default_rng(seed=123)
# Matriz (N, n) donde cada elemento es 1 si el sensor detecta partícula, 0 si no
activaciones_sensores = rng.choice([0, 1], size=(N_simulaciones, n_sensores), p=[1 - prob_sensor, prob_sensor])

# Conteo de sensores activos por evento (suma por fila)
sensores_activos_por_evento = np.sum(activaciones_sensores, axis=1)

# Frecuencia relativa empírica de eventos con exactamente 4 sensores activos
conteo_empirico_k4 = np.sum(sensores_activos_por_evento == k_objetivo)
prob_empirica_k4 = conteo_empirico_k4 / N_simulaciones

# 4. Presentación de resultados y verificación
print(f"=== RESULTADOS DEL RETO DIAGNÓSTICO ===")
print(f"Combinaciones C({n_sensores}, {k_objetivo}): {formas_seleccion} formas distintas de seleccionar los sensores.")
print(f"Probabilidad Teórica P(X = {k_objetivo}): {prob_teorica_k4:.4f} ({prob_teorica_k4*100:.2f}%)")
print(f"Probabilidad Empírica (Monte Carlo): {prob_empirica_k4:.4f} ({prob_empirica_k4*100:.2f}%)")
print(f"Diferencia Absoluta: {abs(prob_empirica_k4 - prob_teorica_k4):.4f}")

# Verificación de tolerancia (el error estocástico debe ser < 0.015 para N=10000)
assert abs(prob_empirica_k4 - prob_teorica_k4) < 0.015, "El error de simulación superó el margen tolerado."
print("\n[OK] Reto Diagnóstico completado exitosamente.")
```
